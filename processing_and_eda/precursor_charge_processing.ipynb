# Combining individually processed files

```python
def combine_parquet_files(directory_path="processed_data", exclude_dir="test_ptm", output_file="combined_data.parquet"):
    """
    Combines all Parquet files in subdirectories of the specified directory into one DataFrame,
    excluding files in the specified directory to exclude.

    Parameters:
    directory_path (str): Path to the main directory containing subdirectories with Parquet files.
    exclude_dir (str): Name of the subdirectory to exclude.
    output_file (str): Name of the output Parquet file to save the combined DataFrame.

    Returns:
    pd.DataFrame: The combined DataFrame.
    """
    all_files = []
    for root, dirs, files in os.walk(directory_path):
        for file in files:
            if file.endswith(".parquet") and exclude_dir not in root:
                all_files.append(os.path.join(root, file))

    combined_df = pd.concat([pd.read_parquet(file, engine="pyarrow") for file in all_files], ignore_index=True)
        
    output_path = os.path.join(directory_path, output_file)
    combined_df.to_parquet(output_path, engine="pyarrow", index=False)
    print(f"Combined data saved to: {output_path}")
    
    return combined_df
```

```python
combined_df = combine_parquet_files(
    directory_path="../processed_data",
    exclude_dir="test_ptm",
    output_file="combined_data_processed.parquet"
)
```

# EDA

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split

def one_hot_to_label(one_hot_vector):
    """Convert a one-hot encoded vector to its corresponding label."""
    return np.argmax(one_hot_vector) + 1

def plot_distribution(dataframe, title, column='charge_label'):
    """Plot the distribution of the specified column."""
    plt.figure(figsize=(12, 6))
    dataframe[column].value_counts().plot(kind='bar')
    plt.title(title)
    plt.xlabel('Charge State')
    plt.ylabel('Frequency')
    plt.show()

def calculate_relative_distribution(dataframe, column='charge_label'):
    """Calculate the relative distribution of the specified column."""
    distribution = dataframe[column].value_counts(normalize=True)
    return distribution * 100

def count_charge_states(df, col):
    """Count unique combinations for charge state vectors."""
    tuples = df[col].apply(lambda x: tuple(x))
    return pd.Series(tuples).value_counts()

def plot_charge_state_vector_distribution(counter, title):
    """Visualize charge state vector distribution."""
    labels, counts = zip(*counter.items())
    label_strings = [str(label) for label in labels]

    plt.figure(figsize=(12, 6))
    plt.bar(range(len(counts)), counts, tick_label=label_strings)
    plt.xlabel("Charge State Vector (binary)")
    plt.ylabel("Count")
    plt.title(title)
    plt.xticks(rotation=90)
    plt.tight_layout()
    plt.show()

def count_and_relative_distribution(df, col):
    """Check for single and multiple charge states."""
    single_charge_state = df[col].apply(lambda x: sum(x) == 1)
    single_count = single_charge_state.sum()
    multiple_count = len(single_charge_state) - single_count
    
    total_count = len(single_charge_state)
    single_relative = single_count / total_count
    multiple_relative = multiple_count / total_count
    
    return single_count, multiple_count, single_relative, multiple_relative

def plot_relative_distributions(single_rel, multiple_rel, title, ax):
    """Plot the relative distribution of single vs. multiple charge states."""
    labels = ['Single Charge State', 'Multiple Charge States']
    relative_counts = [single_rel, multiple_rel]
    ax.bar(labels, relative_counts, color=['C0', 'C1'])
    ax.set_ylabel("Relative Distribution")
    ax.set_title(title)
    ax.set_ylim(0, 1)

def plot_mean_charge_state_distribution(df):
    """Plot the mean charge state distribution per charge state vector."""
    # Group by charge state vector string
    df['charge_state_vector_str'] = df['observed_charge_states'].apply(lambda x: str(x))
    grouped = df.groupby('charge_state_vector_str')

    # Calculate mean of charge state distribution
    agg_intensity_dist_df = grouped['charge_state_dist'].apply(
        lambda x: np.mean(np.vstack(x), axis=0)).reset_index(drop=True)

    # Extract unique charge state vectors
    unique_charge_state_vectors = grouped['observed_charge_states'].first().apply(pd.Series).reset_index(drop=True)
    unique_charge_state_vectors.columns = [i for i in range(1, 7)]

    # Convert aggregated intensity distribution to DataFrame
    agg_intensity_dist_df = pd.DataFrame(agg_intensity_dist_df.tolist(), columns=[i for i in range(1, 7)])

    # Charge state vectors as labels
    ytick_labels = [str(vec) for vec in grouped['observed_charge_states'].first().index]

    # Function to sort charge state vectors
    def sort_charge_state_vectors(row):
        values = row.values
        num_ones = sum(values)
        order = f"{num_ones:02d}" + "".join(map(str, values))
        return order

    # Sort the DataFrames
    sorted_indices = unique_charge_state_vectors.apply(sort_charge_state_vectors, axis=1).sort_values().index
    unique_charge_state_vectors = unique_charge_state_vectors.loc[sorted_indices]
    agg_intensity_dist_df = agg_intensity_dist_df.loc[sorted_indices]
    ytick_labels = [ytick_labels[i] for i in sorted_indices]

    plt.figure(figsize=(10, 8))

    # Plot mean charge state distribution heatmap
    sns.heatmap(agg_intensity_dist_df, cmap='viridis', cbar=True, yticklabels=ytick_labels)
    plt.title('Mean Charge State Distribution per Charge State Vector')
    plt.xlabel('Charge State')
    plt.ylabel('Charge State Vector')

    plt.tight_layout()
    plt.show()


def split_and_analyze_data(df, test_size=0.1, val_size=0.2222, random_state=42):
    """
    Splits the data into training, validation, and test sets and performs comprehensive EDA on the splits.

    Parameters:
    df (pd.DataFrame): The DataFrame containing the data.
    test_size (float): Proportion of the data to include in the test split.
    val_size (float): Proportion of the training data to include in the validation split.
    random_state (int): Random seed for reproducibility.

    Returns:
    tuple: The training, validation, and test sets as DataFrames.
    """
    # Convert one-hot encoded 'most_abundant_charge_state' to a label for stratification
    df['charge_label'] = df['most_abundant_charge_state'].apply(one_hot_to_label)

    # Split the data into train+val and test sets
    train_val_df, test_df = train_test_split(
        df, test_size=test_size, stratify=df['charge_label'], random_state=random_state
    )

    # Split the remaining data into train and validation sets
    train_df, val_df = train_test_split(
        train_val_df, test_size=val_size, stratify=train_val_df['charge_label'], random_state=random_state
    )

    # EDA for charge state distribution
    print("\nTrain Set Relative Distribution:")
    print(calculate_relative_distribution(train_df, 'charge_label'))

    print("\nValidation Set Relative Distribution:")
    print(calculate_relative_distribution(val_df, 'charge_label'))

    print("\nTest Set Relative Distribution:")
    print(calculate_relative_distribution(test_df, 'charge_label'))

    plot_distribution(train_df, 'Training Set Distribution', column='charge_label')
    plot_distribution(val_df, 'Validation Set Distribution', column='charge_label')
    plot_distribution(test_df, 'Test Set Distribution', column='charge_label')

    # EDA for charge state vectors
    print("\nCharge State Vector Distribution - Training Set:")
    train_counter = count_charge_states(train_df, 'observed_charge_states')
    plot_charge_state_vector_distribution(train_counter, "Training Set Charge State Vector Distribution")

    print("\nCharge State Vector Distribution - Validation Set:")
    val_counter = count_charge_states(val_df, 'observed_charge_states')
    plot_charge_state_vector_distribution(val_counter, "Validation Set Charge State Vector Distribution")

    print("\nCharge State Vector Distribution - Test Set:")
    test_counter = count_charge_states(test_df, 'observed_charge_states')
    plot_charge_state_vector_distribution(test_counter, "Test Set Charge State Vector Distribution")

    # Single vs. Multiple charge states analysis
    fig, axs = plt.subplots(1, 3, figsize=(18, 6))
    train_single_rel, train_multiple_rel = count_and_relative_distribution(train_df, 'observed_charge_states')[2:]
    plot_relative_distributions(train_single_rel, train_multiple_rel, "Training Set", axs[0])

    val_single_rel, val_multiple_rel = count_and_relative_distribution(val_df, 'observed_charge_states')[2:]
    plot_relative_distributions(val_single_rel, val_multiple_rel, "Validation Set", axs[1])

    test_single_rel, test_multiple_rel = count_and_relative_distribution(test_df, 'observed_charge_states')[2:]
    plot_relative_distributions(test_single_rel, test_multiple_rel, "Test Set", axs[2])

    plt.tight_layout()
    plt.show()

    # Plot mean charge state distribution heatmap for each set
    print("\nMean Charge State Distribution Heatmap - Training Set:")
    plot_mean_charge_state_distribution(train_df)

    print("\nMean Charge State Distribution Heatmap - Validation Set:")
    plot_mean_charge_state_distribution(val_df)

    print("\nMean Charge State Distribution Heatmap - Test Set:")
    plot_mean_charge_state_distribution(test_df)

    return train_df, val_df, test_df

```
